```python
import os, shutil

# Define paths using your exact Kaggle directories
CODE = '/kaggle/input/datasets/abdullahsirajkhan/mk-code'
DATA = '/kaggle/input/datasets/abdullahsirajkhan/mk-dataset/data/ClinicDB'
WORK = '/kaggle/working/mkunet'

# 1. Copy necessary files to writable working directory
os.makedirs(WORK, exist_ok=True)
for f in ['mkunet_network.py', 'train_polyp.py', 'test_polyp.py']:
    shutil.copy(os.path.join(CODE, f), WORK)
shutil.copytree(os.path.join(CODE, 'utils'), os.path.join(WORK, 'utils'), dirs_exist_ok=True)

# 2. Link dataset into the directory structure the code expects
target = os.path.join(WORK, 'data/polyp/target/ClinicDB')
os.makedirs(os.path.dirname(target), exist_ok=True)
if os.path.islink(target) or os.path.exists(target):
    os.remove(target)
os.symlink(DATA, target)

# 3. Patch train_polyp.py to only run ONE time instead of FIVE times
train_file = os.path.join(WORK, 'train_polyp.py')
with open(train_file, 'r') as f:
    content = f.read()
content = content.replace('for run in [1,2,3,4,5]:', 'for run in [1]:')
with open(train_file, 'w') as f:
    f.write(content)

print("Setup completed successfully.")
```
